**_BMI_**

```python
print("=" * 7 ,"計算BMI", "=" * 7 ,"\n")
# print用了四個引述值
try:
  weight = float(input("輸入體重(kg)"))
  height = float(input("輸入身高(cm)"))
  if weight <= 0 or height <= 0:
    raise Exception("輸入的數值需大於零")

  bmi = weight / (height/100) ** 2
  #print("\n\n"+"=" * 7+"結果"+"=" * 7+f"\n\n體重:{weight:.1f}\n身高:{height:.1f}\nBMI:{bmi:.2f}")
  #上面這種寫法可讀性不高
  print(f"""

  =======結果=======

  體重:{weight:.1f}
  身高:{height:.1f}
  BMI:{bmi:.2f}
  """)
  if bmi < 18.5:
    print("體重過輕")
  elif bmi < 24:
    print("體重正常")
  elif bmi < 27:
    print("體重過重")
  elif bmi < 30:
    print("輕度肥胖")
  elif bmi < 35:
    print("中度肥胖")
  else:
    print("重度肥胖")
  if not (bmi >=18.5 and bmi < 24):
    max = 24 * (height/100) ** 2
    min = 18.5 * (height/100) ** 2
    print(f"建議體重範圍:{min:.1f}~{max:.1f}")

except ValueError:
  print("輸入內容錯誤")
except Exception as e:
  print(e)
```

- tuple(xxx,xxx,xxx,xxx) ==>群組資料類型

在python標準函示式庫裏面有range
- range(stop) #stop=參數
- range(star,stop,step=1)  
(備註)step=1是有預設值的參數 要更改叫做引述值的呼叫  
如果要寫名稱  就是引述名稱的呼叫




```python
range(10)#代表執行10次:0~9
range(1,11)
range(1,11,2)
```

```python
total = 0
for i in range(1,101):
  total += i
print(total)
```

```python
for i in range(1,10): #巢狀迴圈
  print(f'i={i}')# 外迴圈
  for j in range(1,10):# 內迴圈
    print(f"j={j}",end='')
  print()#預設是換行
```

做九九乘法表

```python
for i in range(1,10):
  for j in range(1,10):
    print(f"{i:<2}*{j:2d} = {i*j:2d}",end="  ")
  print()

```
