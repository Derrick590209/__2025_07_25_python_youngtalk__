```python
#小明想要存錢買一輛機車,機車每輛30000元,他將每月存的錢輸入,當存款足夠買機車時,就顯示提示訊息告知

deposit = 0
while(deposit < 30000):
  input_value = int(input("請輸入本月份的存款"))
  deposit += input_value

print("恭喜!已經存夠了,共存了{deposit}元")
```

```python
#小明想要存錢買一輛機車,機車每輛30000元,他將每月存的錢輸入,當存款足夠買機車時,就顯示提示訊息告知

deposit = 0
count = 0
while deposit < 30000:
  count += 1
  input_value = int(input("請輸入本月份的存款"))
  deposit += input_value

print(f"恭喜!已經存了{count}月,共存了{deposit}元")
```

# break:直接跳出迴圈

```python
#小明想要存錢買一輛機車,機車每輛30000元,他將每月存的錢輸入,當存款足夠買機車時,就顯示提示訊息告知

deposit = 0
count = 0
while True: #無限迴圈
  count += 1
  input_value = int(input("請輸入本月份的存款"))
  deposit += input_value
  if deposit >= 30000:
    break

print(f"恭喜!已經存了{count}月,共存了{deposit}元")
```
