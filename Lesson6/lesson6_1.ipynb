```python

while(True):
  try:
   height = int(input("請輸入身高(100~250);"))
   if height < 100 or height > 250:
    print("範圍錯誤，請重新輸入")
    continue
   else:
    break
  except ValueError:
    print("格式錯誤!請重新輸入")
  except Exception as e:
    print(f"{e},請重新輸入")

while(True):
  try:
   weight = int(input("請輸入體重(10~200);"))
   if weight < 10 or weight > 200:
    print("範圍錯誤，請重新輸入")
    continue
   else:
    break
  except ValueError:
    print("格式錯誤!請重新輸入")
  except Exception as e:
    print(f"{e},請重新輸入")


bmi = weight / (height/100)**2
print(f"身高:{height},體重:{weight}BMI為:{bmi:.2f}")
if bmi < 18.5:
  print("體重過輕")
elif bmi < 24:
  print("健康")
elif bmi < 27:
  print("體重過重")
elif bmi < 30:
  print("輕度肥胖")
elif bmi < 35:
  print("中度肥胖")
else:
  print("重度肥胖")
min = 18.5 *  (height/100)**2
max = 24 * (height/100)**2
print(f"標準體重為{min:.2f}~{max:.2f}")
print("應用程式結束")
```

# random回傳整數的函式
1.random.randomrange(stop)==>stop:參數  
2.random.randomrange(start, stop[, step]) ==>[]可有可無  
3.random.randint(a,b)==包含b

```python
import random
random.randrange(5)#5==引數值 不包含5
random.randrange(1,3)#==不包含3
random.randint(1,3) #==包含3
```

```python
#猜數字遊戲
import random

print("======猜數字遊戲======")
min = 1
max = 100
target = random.randint(min,max)
print(target)
while True:
  #input("猜數字的範圍:",min,"~",max) # input只能有一個引數值 所以不對 更新如下
  keyin = int(input(f"猜數字的範圍:{min}~{max}:"))
  if keyin >= min and keyin <= max:
    if target == keyin:
      print(f"賓果!猜對了,答案是:{target}")
      break
  else:
    print("請輸入提示範圍內的數字")
```
