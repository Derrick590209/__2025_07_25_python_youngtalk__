```python
# 學生總分為300
# 要檢查學生輸入的分數,是否是0~300
# 有些學生可以加分5%
# 學生分數加分後,不可以超過300

scores = int(input("請輸入學生分數(最高300分):"))
if scores >=0 and scores <=300:
  is_add = input("學生是否符合加分條件(y,n)")

  # 巢狀判斷 處理兩個條件
  if is_add == "y":
    scores *=1.05
    if scores > 300:
      scores = 300

  print(round(scores))
else:
  print("輸入的範圍,必需在0~300之間")

print("應用程式結束")
```

常見的（至少）兩種不同的錯誤類別為：語法錯誤 (syntax error) 和例外 (exception)。

# Error的種類
1. ZeroDivisionError
2. NameError
3. TypeError

```python
try:
  raise TypeError
# TypeError是內建的 所以可以自己reaise Error
except: # 任何的錯誤
  print("出錯了")

print("應用程式結束")
```

```python
try:
  raise TypeError
except TypeError:#指定TypeError
  print("資料類型出錯了")

print("應用程式結束")
```

```python
try: #說明except可以很多個
  raise ValueError
except TypeError: #指定錯誤
  print("資料類型出錯了")
except ValueError: #指定錯誤
  print("值出錯了")
except:#代表任何錯誤
  print("出錯了")

print("應用程式結束")
```

```python
try:
  raise NameError
except TypeError: #指定錯誤
  print("資料類型出錯了")
except ValueError: #指定錯誤
  print("值出錯了")
except:
  print("出錯了")

print("應用程式結束")
```

- 可以多個except程式區塊
- except程式區塊  只執行一個後  跳出try_except
- except(所有錯誤, 一定要放在最後面)

```python
try: #把except放到前面來的結果
  raise NameError
except:
  print("出錯了")

except TypeError: #指定錯誤
  print("資料類型出錯了")
except ValueError: #指定錯誤
  print("值出錯了")

print("應用程式結束")
```

```python
try:
  raise NameError("不可以除以0")#()內就是錯誤的描述
except TypeError: #指定錯誤
  print("資料類型出錯了")
except ValueError: #指定錯誤
  print("值出錯了")
except Exception as e:#代表所有錯誤的另一種寫法
  print(f"出錯了:{e}")

print("應用程式結束")
```
