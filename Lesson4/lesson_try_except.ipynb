# try...except 是 Python 用來**處理錯誤（例外 Exception）**的語法。  
當程式發生錯誤時，如果沒有處理，程式會直接停止。



```python
try:
    可能發生錯誤的程式
except:
    發生錯誤時執行的程式
```

````python
try:
  age = int(input("請輸入年紀:"))
  print(f"你輸入的年紀是{age}歲")
except:
  print(f"輸入的年紀錯誤")
````

# 常見錯誤種類
## ValueError

```python
int("abc")
錯誤:
ValueError
```
## ZeroDivisionError
```python
10 / 0
錯誤：
ZeroDivisionError
```
## NameError
```python
print(x)
錯誤：
NameError
```




# 多個except

````python
try:
    num = int(input("輸入數字："))
    result = 10 / num
    print(result)

except ValueError:
    print("請輸入數字")

except ZeroDivisionError:
    print("不能除以0")
````

````python
try:
  num = int(input("輸入數字"))
  print(num)

except Exception as a:
  print(f"錯誤內容:{a}")

print("程式結束")
````
