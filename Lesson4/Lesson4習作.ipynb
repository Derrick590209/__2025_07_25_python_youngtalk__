- 體重:90

- 狀態:輕度肥胖
- 建議您的體重:XXX,可以達到最佳狀態


```python
try:
  heigh = float(input("請輸入您的身高:"))
  weigh = float(input("請輸入您的體重:"))
  bmi =  weigh / (heigh/100)**2
  mid_weight = (heigh - 80)*0.7

  print(f"您的身高為{heigh}公分,體重為{weigh}公斤,BMI為:{bmi:.2f}")

  if bmi < 18.5:
    print("體重過輕,需要多吃一點")
  elif 18.5 <= bmi < 24:
    print("健康體重,要繼續保持")
  elif 24 <= bmi < 27:
    print("體重過重,要小心了")

  else:
    print("太胖了,需要馬上減肥")

except ValueError:
  print("請輸入數字")

finally:
  print(f"你的理想體重是:{mid_weight:.2f}")
```
