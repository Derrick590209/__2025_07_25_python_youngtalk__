```python
#讓使用者輸入梯形的上底、下底及高，程式會計算梯形的面積(上底加下底乘以高除以2)
top = float(input("請輸入上底:"))
bottom = float(input("請輸入下底:"))
height = float(input("請輸入高:"))
area = (top + bottom) * height / 2

print("\n======答案是=========",end="\n\n")
print("上底:", top,sep="")
print("下底:", bottom,sep="")
print("高:", height,sep="")
print("梯形的面積:", round(area,ndigits=2),sep="")
```

```python
# 請以(複合指定運算子)程式設計,讓使用者任意輸入三個數字,程式會顯示三數相加的總和(float)
total = 0
number = input("請輸入第一個數字:")
total += float(number)
number = input("請輸入第二個數字:")
total += float(number)
number = input("請輸入第三個數字:")
total += float(number)
print("三數相加總和為:",round(total, ndigits = 2))
```
