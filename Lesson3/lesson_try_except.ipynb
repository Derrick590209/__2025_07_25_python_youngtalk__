# try...except 是 Python 用來**處理錯誤（例外 Exception）**的語法。  
當程式發生錯誤時，如果沒有處理，程式會直接停止。



```python
try:
    可能發生錯誤的程式
except:
    發生錯誤時執行的程式
```

````python
try:
  age = int(input("請輸入年紀:"))
  print(f"你輸入的年紀是{age}歲")
except:
  print(f"輸入的年紀錯誤")
````

# 常見錯誤種類
1.ValueError

```python
int("abc")
錯誤:
ValueError
```
2.ZeroDivisionError
```python
10 / 0
錯誤：
ZeroDivisionError
```
3.NameError
```python
print(x)
錯誤：
NameError
```




# 多個except

````python
try:
    num = int(input("輸入數字："))
    result = 10 / num
    print(result)

except ValueError:
    print("請輸入數字")

except ZeroDivisionError:
    print("不能除以0")
````

# except Exception as e:  
可以把它當成一個需要記憶的固定語法。  
你可以拆成三部分理解：

except # 捕捉錯誤  
Exception # 錯誤類型  
as e  # 把錯誤物件存到e



````python
try:
    x = 10 / 0

except Exception as e:
    print("發生錯誤：", e)
````

````python
try:
  num = int(input("輸入數字"))
  print(num)

except Exception as e:
  print(f"錯誤內容:{e}")

print("程式結束")
````
